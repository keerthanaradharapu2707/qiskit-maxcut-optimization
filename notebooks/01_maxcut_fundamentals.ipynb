# Max-Cut fundamentals
This notebook uses the project package to parse a graph, visualize it, and calculate an exact classical Max-Cut baseline. The default graph is a small demonstration instance.

```python
from maxcut_benchmark.graph_builder import parse_edge_list
from maxcut_benchmark.objective import cut_value, exact_maxcut
import networkx as nx
import matplotlib.pyplot as plt

graph = parse_edge_list('0-1, 1-2, 2-3, 3-0, 0-2')
optimum, bits = exact_maxcut(graph)
print('Vertices:', graph.number_of_nodes(), 'Edges:', graph.number_of_edges())
print('Exact cut value:', optimum, 'Bitstring (vertex 0 first):', bits)
pos = nx.spring_layout(graph, seed=7)
nx.draw_networkx(graph, pos=pos, node_color='#cfe8ff')
plt.axis('off'); plt.show()
```
