# QAOA implementation
This notebook optimizes a small QAOA Max-Cut circuit using a statevector objective, then samples the resulting measured circuit on the local Aer simulator.

```python
from maxcut_benchmark.graph_builder import parse_edge_list
from maxcut_benchmark.qaoa_solver import optimize_qaoa
from maxcut_benchmark.execution import run_local_simulator
from maxcut_benchmark.metrics import metrics_from_counts
from maxcut_benchmark.objective import exact_maxcut

graph = parse_edge_list('0-1, 1-2, 2-3, 3-0, 0-2')
sol = optimize_qaoa(graph, p=1, maxiter=50, seed=42)
counts = run_local_simulator(sol['circuit'], shots=1024, seed=42)
optimum, _ = exact_maxcut(graph)
print('Optimized expected cut:', sol['optimized_expected_cut'])
print(metrics_from_counts(graph, counts, optimum))
```
